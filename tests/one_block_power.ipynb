```python
# p_limits = p_limits_orig.copy()
# print(p_limits_orig)
# p_limits[0] = 1.5
# p_limits[1] = 1.5
# current_max = 0
# for i in range(5):
#     if p_limits[i] < current_max:
#         p_limits[i] = current_max
#     else:
#         current_max = p_limits[i]
# print(p_limits)
def are_p_limits_posible(batt, p_limits, max_soc = 1.):
    """
   Determine if the p_limits are possible to achieve with the battery
    Args:
    --------
    batt: BatteryStorage object
    p_limits: list of floats, limit powers for every block
    """
    
    print(p_limits)
    for _, df_tmp in batt.results.iterrows():
        # lower the peak
        p_limit = p_limits[int(df_tmp.block)-1]
        if df_tmp.p > p_limit:
            # if battery is not empty
            if batt.current_e_kwh >= 0:
                if df_tmp.p - p_limit > batt.max_discharge_p_kw:
                    # not enough max output
                    return(-1)
                # Lower te peak till p_limit
                needed_output = float(df_tmp.p - p_limit)
                if batt.current_e_kwh < needed_output * 0.25:
                    # not enough energy
                    return(-1)
                batt.discharge_amount = needed_output
                batt.discharge(batt.discharge_amount, 0.25)
            else:
                # we have an empty battery
                return(-1)
        else:
            # excess power to charge the battery
            diff = p_limit - df_tmp.p
            batt.charge_amount = float(min(diff, batt.max_charge_p_kw))
            if batt.current_e_kwh < batt.max_e_kwh*max_soc:
                if batt.current_e_kwh + batt.charge_amount * 0.25 > batt.max_e_kwh*max_soc:
                    batt.charge_amount = float(batt.max_e_kwh*max_soc -
                                               batt.current_e_kwh) * 4
                batt.charge(batt.charge_amount, 0.25)
            else:
                pass
    return(1)
```

```python
block = 1
def find_block_p_limit(batt, p_limits, block, p_limits_orig= None):

    max_bound = p_limits[block-1]
    min_bound = p_limits_orig[block-1]- batt.max_discharge_p_kw-1
    if block == 1:
        function = lambda x: are_p_limits_posible(batt, [x if i == block - 1 else p_limits[i] for i in range(5)], max_soc=0.95)
    else:
        function = lambda x: are_p_limits_posible(batt, [x if i == block - 1 else p_limits[i] for i in range(5)])
    root = optimize.bisect(function,
                           min_bound,
                           max_bound,
                           xtol=0.05)
    return(root)
```

```python
def get_max_p_limits(batt):
    """
    Get the maximum possible p_limits for the battery
    Args:
    --------
    batt: BatteryStorage object
    """
    for block in batt.results["block"].unique():
        p_limits = batt.results[batt.results["block"] == block]["p"].max()

    p_limits = [batt.results[batt.results["block"] == block]["p"].max() for block in range(1, 6)]
    #replace nan with 0
    p_limits_orig = [0 if np.isnan(x) else x for x in p_limits]
    
    p_limits = p_limits_orig.copy()
    current_max = 0
    for i in range(5):
        if p_limits[i] < current_max:
            p_limits[i] = current_max
        else:
            current_max = p_limits[i]
    return p_limits, p_limits_orig
def find_p_limits(batt):
    """
    Find the optimal p_limits for the battery
    Args:
    --------
    batt: BatteryStorage object
    p_limits: list of floats, limit powers for every block
    """
    p_limits, p_limits_orig = get_max_p_limits(batt)
    print(p_limits_orig)
    for block in range(1, 6):
        if block == 1:
            p_limit = round(find_p_limit(batt, p_limits, block, p_limits_orig) + 0.1, 1)
            p_limits[0] = p_limit
        else:
            p_limits_min = p_limits.copy()
            p_limits_min[block-1] = p_limits[block-2]
            if are_p_limits_posible(batt, p_limits_min) == 1:
                p_limits[block-1] = p_limits[block-2]
            else:
                p_limit = round(find_p_limit(batt, p_limits, block, p_limits_orig) + 0.1, 1)
                p_limits[block-1] = p_limit
    
    return(p_limits)
    
```

```python
p_limits = find_p_limits(batt)
lst = []
batt.hard_reset()
for key, df_tmp in batt.results.iterrows():
    p_limit = p_limits[int(df_tmp.block)-1]
    if df_tmp.p > p_limit:
        if batt.current_e_kwh > 0:
            # if battery is not empty
            batt.discharge_amount = float(
                min(df_tmp.p - p_limit, batt.max_discharge_p_kw))
            if batt.current_e_kwh < batt.discharge_amount * 0.25:
                batt.discharge_amount = batt.current_e_kwh * 4
            batt.discharge(batt.discharge_amount, 0.25)
            batt.results.loc[
                key, "battery_plus"] = batt.discharge_amount
        else:
            # we have an empty battery
            pass
    # charging battery
    else:
        excess_power = p_limit - df_tmp.p
        if batt.current_e_kwh < batt._max_e_kwh:
            batt.charge_amount = min(excess_power,
                                     batt.max_charge_p_kw)
            if batt.current_e_kwh + batt.charge_amount * 0.25 > batt._max_e_kwh:
                batt.charge_amount = (batt.max_e_kwh -
                                      batt.current_e_kwh) * 4
            batt.results.loc[key,
                             "battery_minus"] = -batt.charge_amount
            batt.charge(batt.charge_amount, 0.25)
        else:
            # we have a full battery
            pass
    lst.append(batt.current_e_kwh)
```

```python
batt.results[
            "p_after"] = batt.results.p - batt.results.battery_plus - batt.results.battery_minus
batt.results["var_bat"] = lst
```
